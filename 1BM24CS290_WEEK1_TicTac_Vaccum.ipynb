```python

# Tic-Tac-Toe Game

board = [" " for i in range(9)]

def display_board():
    print()
    print(board[0], "|", board[1], "|", board[2])
    print("--+---+--")
    print(board[3], "|", board[4], "|", board[5])
    print("--+---+--")
    print(board[6], "|", board[7], "|", board[8])
    print()

def check_winner(player):
    winning_positions = [
        [0, 1, 2],
        [3, 4, 5],
        [6, 7, 8],
        [0, 3, 6],
        [1, 4, 7],
        [2, 5, 8],
        [0, 4, 8],
        [2, 4, 6]
    ]

    for position in winning_positions:
        if (board[position[0]] == player and
            board[position[1]] == player and
            board[position[2]] == player):
            return True

    return False

def tic_tac_toe():
    current_player = "X"

    for turn in range(9):
        display_board()

        print("Player", current_player)

        position = int(input("Enter position (1-9): ")) - 1

        if position < 0 or position > 8 or board[position] != " ":
            print("Invalid position! Try again.")
            continue

        board[position] = current_player

        if check_winner(current_player):
            display_board()
            print("Player", current_player, "wins!")
            return

        if current_player == "X":
            current_player = "O"
        else:
            current_player = "X"

    display_board()
    print("Game Draw!")

tic_tac_toe()
```

```python

# Vacuum Cleaner Agent
# Simple Reflex Agent and Goal-Based Agent

# Initial environment
rooms = {
    "A": "Dirty",
    "B": "Dirty"
}

# Simple Reflex Agent
def simple_reflex_agent():
    print("\n--- SIMPLE REFLEX AGENT ---")

    location = "A"

    while True:
        print("\nCurrent Location:", location)
        print("Room Status:", rooms[location])

        if rooms[location] == "Dirty":
            print("Action: SUCK")
            rooms[location] = "Clean"

        else:
            if location == "A":
                location = "B"
                print("Action: MOVE RIGHT")
            else:
                print("Action: STOP")
                break

    print("\nFinal Room Status:", rooms)


# Reset environment
rooms["A"] = "Dirty"
rooms["B"] = "Dirty"


# Goal-Based Agent
def goal_based_agent():
    print("\n--- GOAL BASED AGENT ---")

    location = "A"
    goal = "All rooms Clean"

    while True:
        print("\nCurrent Location:", location)
        print("Room Status:", rooms[location])

        if rooms[location] == "Dirty":
            print("Action: SUCK")
            rooms[location] = "Clean"

        if all(status == "Clean" for status in rooms.values()):
            print("\nGoal Achieved:", goal)
            break

        if location == "A":
            location = "B"
            print("Action: MOVE RIGHT")


    print("Final Room Status:", rooms)


# Run both agents
simple_reflex_agent()

# Reset rooms before running second agent
rooms["A"] = "Dirty"
rooms["B"] = "Dirty"

goal_based_agent()
```
