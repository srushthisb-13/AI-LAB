```python

# Hill Climbing Algorithm for 4-Queens Problem

N = 4

# Initial configuration
# Each index represents column
# Each value represents row
state = [1, 1, 1, 1]


# Function to calculate cost
def calculate_cost(state):
    conflicts = 0

    for i in range(N):
        for j in range(i + 1, N):

            # Same row or same diagonal
            if (state[i] == state[j] or
                abs(state[i] - state[j]) == abs(i - j)):

                conflicts += 1

    return conflicts


# Generate neighbouring states
def generate_neighbors(state):
    neighbors = []

    for col in range(N):
        for row in range(1, N + 1):

            if row != state[col]:

                new_state = state.copy()
                new_state[col] = row

                neighbors.append(new_state)

    return neighbors


# Hill Climbing Algorithm
def hill_climbing(state):

    step = 0

    while True:

        current_cost = calculate_cost(state)

        print("\nStep:", step)
        print("Current State:", state)
        print("Current Cost:", current_cost)

        # Check goal
        if current_cost == 0:
            print("\nGoal State Reached!")
            break

        neighbors = generate_neighbors(state)

        print("\nNeighbor States and Costs:")

        for neighbor in neighbors:
            cost = calculate_cost(neighbor)
            print(neighbor, "-> Cost:", cost)

        # Select best neighbor
        best_neighbor = min(
            neighbors,
            key=lambda x: (calculate_cost(x), x)
        )

        best_cost = calculate_cost(best_neighbor)

        # Check improvement
        if best_cost >= current_cost:
            print("\nLocal Minimum Reached!")
            break

        state = best_neighbor
        step += 1

    print("\nFinal Configuration:", state)
    print("Final Cost:", calculate_cost(state))


# Execute program
hill_climbing(state)
```
