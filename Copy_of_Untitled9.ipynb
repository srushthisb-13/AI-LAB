```python
def dls(state, goal, limit, path, visited):
    if state == goal:
        return path

    if limit == 0:
        return None

    zero = state.index(0)
    row, col = divmod(zero, 3)

    moves = [
        (-1, 0), (1, 0),
        (0, -1), (0, 1)
    ]

    for dr, dc in moves:
        nr, nc = row + dr, col + dc

        if 0 <= nr < 3 and 0 <= nc < 3:
            new_pos = nr * 3 + nc
            new_state = list(state)

            new_state[zero], new_state[new_pos] = (
                new_state[new_pos], new_state[zero]
            )

            new_state = tuple(new_state)

            if new_state not in visited:
                visited.add(new_state)

                result = dls(
                    new_state, goal, limit - 1,
                    path + [new_state], visited
                )

                if result is not None:
                    return result

                visited.remove(new_state)

    return None


def ids(start, goal, max_depth=50):
    for depth in range(max_depth + 1):
        visited = {start}

        result = dls(
            start, goal, depth, [start], visited
        )

        if result is not None:
            return result

    return None


start = (1, 2, 3,
         4, 0, 6,
         7, 5, 8)

goal = (1, 2, 3,
        4, 5, 6,
        7, 8, 0)

solution = ids(start, goal)

if solution:
    print("Solution found using IDS:")
    for state in solution:
        for i in range(0, 9, 3):
            print(state[i:i+3])
        print()
    print("Number of moves:", len(solution) - 1)
else:
    print("No solution found within the maximum depth.")
```

```python
def dls(state, goal, limit, path):
    if state == goal:
        return path

    if limit == 0:
        return None

    zero = state.index(0)
    r, c = divmod(zero, 3)

    for dr, dc in [(1, 0), (-1, 0), (0, 1), (0, -1)]:
        nr, nc = r + dr, c + dc

        if 0 <= nr < 3 and 0 <= nc < 3:
            pos = nr * 3 + nc
            board = list(state)
            board[zero], board[pos] = board[pos], board[zero]
            new_state = tuple(board)

            if new_state not in path:
                result = dls(
                    new_state, goal, limit - 1,
                    path + [new_state]
                )

                if result is not None:
                    return result

    return None


def ids(start, goal):
    depth = 0

    while True:
        result = dls(start, goal, depth, [start])

        if result is not None:
            return result

        depth += 1


start = (5, 4, 6, 1, 8, 7, 3, 2, 0)
goal = (1, 2, 3, 4, 5, 6, 7, 8, 0)

result = ids(start, goal)

for state in result:
    print(state[:3])
    print(state[3:6])
    print(state[6:])
    print()




```

```python
def dfs(start, goal, limit=30):
    stack = [(start, [], 0)]
    visited = set()

    while stack:
        state, path, depth = stack.pop()

        if state == goal:
            return path + [state]

        if state in visited or depth >= limit:
            continue

        visited.add(state)
        zero = state.index(0)
        row, col = divmod(zero, 3)

        moves = [
            (-1, 0), (1, 0),
            (0, -1), (0, 1)
        ]

        for dr, dc in moves:
            nr, nc = row + dr, col + dc

            if 0 <= nr < 3 and 0 <= nc < 3:
                new_pos = nr * 3 + nc
                new_state = list(state)

                new_state[zero], new_state[new_pos] = (
                    new_state[new_pos], new_state[zero]
                )

                new_state = tuple(new_state)

                if new_state not in visited:
                    stack.append(
                        (new_state, path + [state], depth + 1)
                    )

    return None


start = (1, 2, 3,
         4, 0, 6,
         7, 5, 8)

goal = (1, 2, 3,
        4, 5, 6,
        7, 8, 0)

solution = dfs(start, goal)

if solution:
    print("Solution found using DFS:")
    for state in solution:
        for i in range(0, 9, 3):
            print(state[i:i+3])
        print()
    print("Number of moves:", len(solution) - 1)
else:
    print("No solution found within the depth limit.")
```
