```python

import random
import math

N = 8

# Calculate the number of attacking queen pairs
def calculate_cost(board):
    conflicts = 0

    for i in range(N):
        for j in range(i + 1, N):

            # Same row or same diagonal
            if (board[i] == board[j] or
                abs(board[i] - board[j]) == abs(i - j)):

                conflicts += 1

    return conflicts


# Generate a neighbouring state
def generate_neighbor(board):
    new_board = board.copy()

    # Select a random column
    col = random.randint(0, N - 1)

    # Move queen to another row
    new_row = random.randint(0, N - 1)

    while new_row == new_board[col]:
        new_row = random.randint(0, N - 1)

    new_board[col] = new_row

    return new_board


# Display chessboard
def display_board(board):
    for row in range(N):
        for col in range(N):

            if board[col] == row:
                print(" Q ", end="")
            else:
                print(" . ", end="")

        print()


# Simulated Annealing Algorithm
def simulated_annealing():

    # Random initial configuration
    current = [random.randint(0, N - 1) for _ in range(N)]

    temperature = 100
    cooling_rate = 0.995
    iteration = 0
    max_iterations = 100000

    current_cost = calculate_cost(current)

    print("Initial Board:", current)
    print("Initial Cost:", current_cost)

    while temperature > 0.01 and iteration < max_iterations:

        if current_cost == 0:
            break

        # Generate neighbour
        neighbor = generate_neighbor(current)

        neighbor_cost = calculate_cost(neighbor)

        # Calculate cost difference
        delta = neighbor_cost - current_cost

        # Accept better state
        if delta < 0:
            current = neighbor
            current_cost = neighbor_cost

        # Sometimes accept worse state
        elif random.random() < math.exp(-delta / temperature):
            current = neighbor
            current_cost = neighbor_cost

        # Reduce temperature
        temperature *= cooling_rate

        iteration += 1

        # Display progress
        if iteration % 1000 == 0:
            print("Iteration:", iteration,
                  "Cost:", current_cost,
                  "Temperature:", round(temperature, 2))

    print("\nFinal Configuration:", current)
    print("Final Cost:", current_cost)
    print("Total Iterations:", iteration)

    if current_cost == 0:
        print("\nSolution Found!")
        display_board(current)
    else:
        print("\nSolution Not Found. Try Again.")


# Run program
simulated_annealing()
```
