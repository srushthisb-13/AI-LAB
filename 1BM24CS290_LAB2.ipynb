```python
def dls(state, goal_state, limit, visited):
    print(" " * (5 - limit), state)

    if state == goal_state:
        print("Goal State Found!")
        return True

    if limit == 0:
        return False

    visited.add(state)

    zero = state.index(0)
    row, col = divmod(zero, 3)

    moves = [(-1, 0), (1, 0), (0, -1), (0, 1)]

    for dr, dc in moves:
        nr, nc = row + dr, col + dc

        if 0 <= nr < 3 and 0 <= nc < 3:
            new_zero = nr * 3 + nc
            new_state = list(state)

            new_state[zero], new_state[new_zero] = (
                new_state[new_zero], new_state[zero]
            )

            new_state = tuple(new_state)

            if new_state not in visited:
                visited.add(new_state)

                if dls(new_state, goal_state, limit - 1, visited):
                    return True

                visited.remove(new_state)

    return False


def ids(start_state, goal_state, max_depth=10):
    for depth in range(max_depth + 1):
        print("\nDepth Limit:", depth)
        visited = {start_state}

        if dls(start_state, goal_state, depth, visited):
            return True

    print("Goal not found within maximum depth.")
    return False


start_state = (1, 2, 3,
               4, 0, 6,
               7, 5, 8)

goal_state = (1, 2, 3,
              4, 5, 6,
              7, 8, 0)

ids(start_state, goal_state)
```

```python
def dfs(start, goal, limit=30):
    stack = [(start, [], 0)]
    visited = set()

    while stack:
        state, path, depth = stack.pop()

        if state == goal:
            return path + [state]

        if state in visited or depth >= limit:
            continue

        visited.add(state)
        zero = state.index(0)
        row, col = divmod(zero, 3)

        moves = [
            (-1, 0), (1, 0),
            (0, -1), (0, 1)
        ]

        for dr, dc in moves:
            nr, nc = row + dr, col + dc

            if 0 <= nr < 3 and 0 <= nc < 3:
                new_pos = nr * 3 + nc
                new_state = list(state)

                new_state[zero], new_state[new_pos] = (
                    new_state[new_pos], new_state[zero]
                )

                new_state = tuple(new_state)

                if new_state not in visited:
                    stack.append(
                        (new_state, path + [state], depth + 1)
                    )

    return None


start = (1, 2, 3,
         4, 0, 6,
         7, 5, 8)

goal = (1, 2, 3,
        4, 5, 6,
        7, 8, 0)

solution = dfs(start, goal)

if solution:
    print("Solution found using DFS:")
    for state in solution:
        for i in range(0, 9, 3):
            print(state[i:i+3])
        print()
    print("Number of moves:", len(solution) - 1)
else:
    print("No solution found within the depth limit.")
```
